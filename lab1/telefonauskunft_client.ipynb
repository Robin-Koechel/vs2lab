## Transport

```python
import clientserver
import logging

class Transport:
  def __init__(self):
    self.logger = logging.getLogger("vs2lab.lab1.telefonauskunft_client.Transport") # setup logger
    self.logger.info('logger initialized')
  def send(self, message):
    socket = clientserver.Client() # initialize socket for client
    self.logger.info('client socket initialized')
    socket.call(message) # send message to server via socket
    self.logger.info('message sent')
```

## Auskunft

```python
import logging

class Auskunft:
  def __init__(self):
    self.logger = logging.getLogger('vs2lab.lab1.telefonauskunft_client.Auskunft') # setup logger
    self.transport = Transport()

  def GET(self, name):
    self.logger.info(f'requesting entry with name {name}...')
    self.transport.send(f'GET:{name}') # send request for entry

  def GETALL(self):
    self.logger.info('requesting all entries...')
    self.transport.send('GETALL:') # send request for all entries
```

## Example

```python
auskunft = Auskunft()
print(auskunft.GET("Anna Mueller"))
print(auskunft.GETALL())
```

## Tests

```python
import unittest
from unittest.mock import MagicMock

class ClientTest(unittest.TestCase):
  def setUp(self):
    self.auskunft = Auskunft()
    self.auskunft.transport = MagicMock() # use mock for isolated testing

  def test_should_call_transport_with_name(self):
    self.auskunft.GET('Anna Mueller')
    self.auskunft.transport.send.assert_called_with('GET:Anna Mueller')

  def test_should_call_transport_with_(self):
    self.auskunft.GETALL()
    self.auskunft.transport.send.assert_called_with('GETALL:')

if __name__ == '__main__':
  unittest.main(argv=['ignore-args'], exit=False) # avoid sys.argv args and exiting
```
