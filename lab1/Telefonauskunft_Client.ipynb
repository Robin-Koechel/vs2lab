# Transport Layer - Client

```python
import logging
import clientserver
class Transport:
    def __init__(self):
        self._logger = logging.getLogger("vs2lab.lab1.Telefonauskunft_Client.Transport")
        
        self._logger.info("Client Transport initialized.")
    def send(self, message:str):
        socket = clientserver.Client()
        self._logger.info("Client Transport initialized.")
        return socket.call(message)
```

# Auskunft Layer - Client

```python

import logging


class Phonebook:
    
    def __init__(self):
        self._logger = logging.getLogger("vs2lab.lab1.Telefonauskunft_Client.Auskunft")
        self.transport = Transport()
        
    def GETALL(self):
        self._logger.info("Requesting all entries from phone book.")
        return self.transport.send("GETALL::")

    def GET(self,name:str):
        self._logger.info(f"Requesting entry for {name} from phone book.")
        return self.transport.send(f"GET::{name}")
```

# Beispiel Nutzung - Client

```python
phonebook = Phonebook()
phonebook.GET("Scooby Doo")
phonebook.GET("Shaggy")
phonebook.GETALL()
```

# Unit Tests - Client

```python
import unittest
from unittest.mock import patch, MagicMock

class TestPhonebook(unittest.TestCase):
    def setUp(self):
        # create a Phonebook instance but patch its transport to avoid real sockets
        self.pb = Phonebook()
        self.pb.transport = MagicMock()

    def test_get_calls_transport_with_name(self):
        self.pb.GET('Scooby Doo')
        # ensure transport.send was called with the expected message format
        self.pb.transport.send.assert_called_with('GET::Scooby Doo')

    def test_getall_calls_transport(self):
        self.pb.GETALL()
        self.pb.transport.send.assert_called_with('GETALL::')

class TestIntregration(unittest.TestCase):
    def setUp(self):
        self.phonebook = Phonebook()

    def test_integration_GET(self):
        res1 = self.phonebook.GET("Scooby Doo")
        self.assertEqual(res1,"{'Scooby Doo': '555-1234'}")
        res1 = self.phonebook.GET("Shaggy")
        self.assertEqual(res1,"{'Shaggy': '555-5678'}")
    
    def test_integration_GETALL(self):
        res2 = self.phonebook.GETALL()
        self.assertEqual(res2,"{'Scooby Doo': '555-1234', 'Shaggy': '555-5678'}")
                             

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
